```python
from pathlib import Path
from datetime import datetime

root = Path(r"C:\Mestrado\Graph_Pruning\prodigy")

arquivos = []

for p in root.rglob("*"):
    if p.is_file() and any(
        pasta in {"log", "state", "results", "outputs", "runs"}
        for pasta in p.parts
    ):
        try:
            arquivos.append((p.stat().st_mtime, p))
        except OSError:
            pass

for timestamp, p in sorted(arquivos, reverse=True)[:40]:
    data = datetime.fromtimestamp(timestamp).strftime("%d/%m/%Y %H:%M:%S")
    print(f"{data} | {p}")
```

```python
from pathlib import Path

root = Path(r"C:\Mestrado\Graph_Pruning\prodigy")

arquivos = [
    root / "experiments" / "run_single_experiment.py",
    root / "experiments" / "trainer.py",
]

termos = [
    "eval_only",
    "zero_shot",
    "test_step",
    "test_epoch",
    "evaluate",
    "evaluation",
    "accuracy",
    "metric",
    "wandb.log",
    "save",
    "csv",
]

for arquivo in arquivos:
    print(f"\n{'=' * 80}\nARQUIVO: {arquivo.name}\n{'=' * 80}")

    linhas = arquivo.read_text(encoding="utf-8").splitlines()

    encontrados = set()

    for i, linha in enumerate(linhas):
        if any(termo.lower() in linha.lower() for termo in termos):
            encontrados.update(range(max(0, i - 2), min(len(linhas), i + 4)))

    for i in sorted(encontrados):
        print(f"{i + 1:4}: {linhas[i]}")
```
