Codigo da mochila sem repetição

```python
def knapsack(pesos, valores, capacidade):
    n = len(pesos)

    if n != len(valores):
        raise ValueError("Pesos e valores devem ter o mesmo tamanho.")

    if capacidade < 0:
        raise ValueError("A capacidade não pode ser negativa.")

    if any(p <= 0 for p in pesos):
        raise ValueError("Os pesos devem ser positivos.")

    # dp[i][w] = maior valor usando os primeiros i itens
    # com capacidade w.
    dp = [[0] * (capacidade + 1) for _ in range(n + 1)]

    # Preencher a tabela
    for i in range(1, n + 1):
        peso = pesos[i - 1]
        valor = valores[i - 1]

        for w in range(capacidade + 1):

            # Opção 1: não pegar o item atual
            dp[i][w] = dp[i - 1][w]

            # Opção 2: pegar o item, se ele couber
            if peso <= w:
                valor_com_item = (
                    valor + dp[i - 1][w - peso]
                )

                dp[i][w] = max(
                    dp[i][w],
                    valor_com_item
                )

    # Recuperar os itens escolhidos
    itens_escolhidos = []
    w = capacidade

    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            itens_escolhidos.append(i)
            w -= pesos[i - 1]

    itens_escolhidos.reverse()

    return dp[n][capacidade], itens_escolhidos, dp
```

```python
pesos = [1, 3, 4]
valores = [1, 4, 5]
capacidade = 7

valor_maximo, itens, tabela = knapsack(
    pesos, valores, capacidade
)

print("Valor máximo:", valor_maximo)
print("Itens escolhidos:", itens)
print("Peso total:", sum(pesos[i - 1] for i in itens))
```

Edit Distance

```python
def edit_distance(a, b):
    m = len(a)
    n = len(b)

    # Tabela de (m + 1) linhas e (n + 1) colunas
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    # Casos-base: transformar uma string em uma string vazia
    for i in range(m + 1):
        dp[i][0] = i

    # Casos-base: transformar uma string vazia em outra string
    for j in range(n + 1):
        dp[0][j] = j

    # Preencher a tabela
    for i in range(1, m + 1):
        for j in range(1, n + 1):

            # Se os caracteres são iguais, não há custo adicional
            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]

            else:
                # Remoção, inserção ou substituição
                dp[i][j] = 1 + min(
                    dp[i - 1][j],      # remoção
                    dp[i][j - 1],      # inserção
                    dp[i - 1][j - 1]   # substituição
                )

    return dp[m][n]


# Exemplo
print(edit_distance("gato", "miau"))  # Saída: 3
```

Fibonnacci

```python
def fibonacci(n):
    if n < 0:
        raise ValueError("n deve ser não negativo")

    if n <= 1:
        return n

    dp = [0] * (n + 1)

    dp[0] = 0
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
```

```python
print(fibonacci(10))
```

Cutting rod

```python
def rod_cutting(prices, n):
    # prices[i] é o preço de uma barra de comprimento i.
    # Usamos prices[0] = 0 para facilitar os índices.
    if n < 0 or n >= len(prices):
        raise ValueError("O comprimento n não é válido para a tabela de preços.")

    # best[j] = maior receita para uma barra de comprimento j
    best = [0] * (n + 1)

    # first_cut[j] = tamanho do primeiro pedaço escolhido
    first_cut = [0] * (n + 1)

    # Calculamos a solução para comprimentos crescentes
    for length in range(1, n + 1):
        max_revenue = float("-inf")

        # Experimentamos todos os tamanhos para o primeiro pedaço
        for cut in range(1, length + 1):
            revenue = prices[cut] + best[length - cut]

            if revenue > max_revenue:
                max_revenue = revenue
                first_cut[length] = cut

        best[length] = max_revenue

    # Recuperamos os cortes da solução ótima
    cuts = []
    remaining = n

    while remaining > 0:
        cut = first_cut[remaining]
        cuts.append(cut)
        remaining -= cut

    return best[n], cuts, best
```

```python
prices = [0, 1, 5, 8, 9]

revenue, cuts, best = rod_cutting(prices, 4)

print("Receita máxima:", revenue)
print("Comprimentos dos pedaços:", cuts)
print("Melhores receitas:", best)
```

Assembly Line Scheduling

```python
def assembly_line_scheduling(a1, a2, t1, t2, e1, e2, x1, x2):
    n = len(a1)

    # Verificar se os vetores têm tamanhos compatíveis
    if n == 0 or len(a2) != n:
        raise ValueError("As duas linhas devem ter o mesmo tamanho e não podem ser vazias.")

    if len(t1) != n - 1 or len(t2) != n - 1:
        raise ValueError("Os vetores de transferência devem ter n-1 elementos.")

    # f1[j] e f2[j]: menor tempo para chegar à estação j
    # terminando, respectivamente, na linha 1 ou na linha 2.
    f1 = [0] * n
    f2 = [0] * n

    # Guardam a linha anterior escolhida para chegar à estação j
    l1 = [0] * n
    l2 = [0] * n

    # Inicialização: primeira estação
    f1[0] = e1 + a1[0]
    f2[0] = e2 + a2[0]

    # Preencher as tabelas, estação por estação
    for j in range(1, n):

        # Para terminar a estação j na linha 1:
        ficar_linha1 = f1[j - 1] + a1[j]
        trocar_para_linha1 = f2[j - 1] + t2[j - 1] + a1[j]

        if ficar_linha1 <= trocar_para_linha1:
            f1[j] = ficar_linha1
            l1[j] = 1
        else:
            f1[j] = trocar_para_linha1
            l1[j] = 2

        # Para terminar a estação j na linha 2:
        ficar_linha2 = f2[j - 1] + a2[j]
        trocar_para_linha2 = f1[j - 1] + t1[j - 1] + a2[j]

        if ficar_linha2 <= trocar_para_linha2:
            f2[j] = ficar_linha2
            l2[j] = 2
        else:
            f2[j] = trocar_para_linha2
            l2[j] = 1

    # Escolher a melhor linha para sair da fábrica
    sair_linha1 = f1[n - 1] + x1
    sair_linha2 = f2[n - 1] + x2

    if sair_linha1 <= sair_linha2:
        tempo_minimo = sair_linha1
        linha_atual = 1
    else:
        tempo_minimo = sair_linha2
        linha_atual = 2

    # Reconstruir o caminho de trás para frente
    caminho = [0] * n
    caminho[n - 1] = linha_atual

    for j in range(n - 1, 0, -1):
        if caminho[j] == 1:
            caminho[j - 1] = l1[j]
        else:
            caminho[j - 1] = l2[j]

    return tempo_minimo, caminho, f1, f2
```

```python
# Tempo de processamento das estações
a1 = [7, 9, 3, 4, 8, 4]  # Linha 1
a2 = [8, 5, 6, 4, 5, 7]  # Linha 2

# Tempos de transferência entre as estações
t1 = [2, 3, 1, 3, 4]  # Linha 1 -> Linha 2
t2 = [2, 1, 2, 2, 1]  # Linha 2 -> Linha 1

# Tempos de entrada
e1 = 2
e2 = 4

# Tempos de saída
x1 = 3
x2 = 2
```

```python
tempo, caminho, f1, f2 = assembly_line_scheduling(
    a1, a2, t1, t2, e1, e2, x1, x2
)

print("Menor tempo:", tempo)
print("Linha escolhida por estação:", caminho)
print("Custos acumulados na linha 1:", f1)
print("Custos acumulados na linha 2:", f2)
```
